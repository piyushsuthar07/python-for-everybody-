```python
class Solution1:
    def pattern7(self, n):
        for i in range(n):
            for j in range(1,n-i):
                print(" ", end="")
            for k in range(2*i+1):
                print("*",end="")
            print()

class Solution2:
    def pattern8(self, n):
        for i in range(n):
            for j in range(i):
                print(" ", end="")
            for k in range(2*n-(2*i+1)):
                print("*",end="")
            print()
solution1 = Solution1()
solution1.pattern7(5)
solution2 = Solution2()
solution2.pattern8(5)
```
