```python
class Solution:
    def pattern12(self, n):
        for i in range(n):
            for j in range(i+1):
                print("*",end="")
            print()
        for k in range(1, n):
            for l in range(n-k):
                print("*", end= "")
            print()
solution =Solution()
solution.pattern12(5)
```

```python
class Solution:
    def pattern13(self, n):
        for i in range(n):
            for j in range(i+1):
                if (i+j)%2 == 0:
                    print(1, end=" ")
                else:
                    print(0, end = " ")
            print()
solution =Solution()
solution.pattern13(5)
```

```python
class Solution:
    def pattern14(self, n):
        for i in range(1,n+1):
            for j in range(1, i+1):
                print(j, end="")
            
            for j in range(1, 2*(n-i)+1):
                print(" ", end= "")
            
            for j in range(i,0,-1):
                print(j,end="")
            print()
solution =Solution()
solution.pattern14(5)
```

```python
class Solution:
    def pattern15(self,n):
        point = 1
        for i in range(1,n+1):
            for j in range(1,i+1):
                print(point, end= " ")
                point+=1
            print()

solution =Solution()
solution.pattern15(5)
```

```python

class Solution:
    def pattern16(self, n):
        for i in range(1, n + 1):
            for j in range(1, i + 1):
                print(chr(64 + j), end="")
            print()

solution = Solution()
solution.pattern16(5)
```
