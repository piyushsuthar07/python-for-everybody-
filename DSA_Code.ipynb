Find Min & Max

```python
def find_max(arr):
    Maximum = arr[0]

    for num in arr:
        if num > Maximum:
            Maximum = num
    
    return Maximum

print(find_max([4, 8, 3, 9, 1, 6]))
```

```python
def find_min(arr):
    minimum = arr[0]

    for num in arr:
        if num < minimum:
            minimum = num

    return minimum

print(find_min([4, 8, 3, 9, 1, 6]))
```
