Find Min & Max

```python
def find_max(arr):
    Maximum = arr[0]

    for num in arr:
        if num > Maximum:
            Maximum = num
    
    return Maximum

print(find_max([4, 8, 3, 9, 1, 6]))
```

```python
def find_min(arr):
    minimum = arr[0]

    for num in arr:
        if num < minimum:
            minimum = num

    return minimum

print(find_min([4, 8, 3, 9, 1, 6]))
```

```python
def Find_Second_Large(arr):
    Largest = arr[0]
    Second_Large = arr[1]

    if Second_Large > Largest:
        Largest,Second_Large = Second_Large,Largest
    for current in arr[2:]:
        if current>Largest:
            Second_Large = Largest
            Largest = current
        elif current>Second_Large:
            Second_Large = current
    return Second_Large, Largest

print(Find_Second_Large([12,43,54,23,65,45,34,54,23]))

```

<h2> Shifting Zero</h2>

```python
def move_zeros(arr):
    position = 0

    # Place all non-zero elements at the front
    for current in arr:
        if current != 0:
            arr[position] = current
            position += 1

    #remaining positions with zeros
    while position < len(arr):
        arr[position] = 0
        position += 1

    return arr
print(move_zeros([0, 1, 0, 3, 12, 0, 33, 12]))
```

<h2> Remove Duplicates </h2>

```python
def remove_Duplicate(arr):
    position = 0

    for current in arr:
        if current != arr[position]:
            position += 1
            arr[position] = current

    return arr[:position + 1]

print(remove_Duplicate([1,1,3,3,4,5]))
```
